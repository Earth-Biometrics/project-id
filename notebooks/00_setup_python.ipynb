# Project Setup (Python & libraries)

For MacOS:
This notebook provides a step-by-step setup for the Python venv and libraries to support:

- `third-party/eb_megadetector`
- `third-party/bioclip-2`

(For Spark Station, run 'bash setup_spark_station.sh')

## Step 1 — Create and activate a Python environment

Use one of the options below.


#### Option A (venv)
```bash
python3.13 -m venv .venv313
source .venv313/bin/activate
# if needed:
python -m pip install --upgrade pip
```
#### Option B (vu) 
```bash
pip install --user uv
export PATH="$HOME/.local/bin:$PATH"
# verify
uv --version
# install venv with uv
uv venv .venv313 --python 3.13
source .venv313/bin/activate
```
#### Option C (conda)
```bash
conda create -n earth-biometrics python=3.13 -y
conda activate earth-biometrics
python -m pip install --upgrade pip
```


## Step 2 — Install Python libs

```bash
source .venv/bin/activate
pip install --upgrade pip
pip install -r requirements.txt
```

## Appendix of misc Python environments

```bash
sudo add-apt-repository ppa:deadsnakes/ppa
sudo apt update

sudo apt install python3.13 python3.13-venv python3.13-dev
pip freeze > requirements.txt
python3.13 -m venv venv313
source venv313/bin/activate
pip install -r requirements.txt

# if frequently need to toggle between Python 3.11 and 3.13, use pyenv:
pyenv install 3.13.0
pyenv global 3.13.0

```
